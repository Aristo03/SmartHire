## 2. Dataset Distribution and Quality Analysis

Investigating the distribution of resume categories,
missing values and duplicate records.

```python
resume_df["Category"].value_counts()
```

```python
print("Total categories:", resume_df["Category"].nunique())
```

```python
resume_df["Category"].value_counts().describe()
```

```python
category_counts = resume_df["Category"].value_counts()

plt.figure(figsize=(12, 6))

category_counts.plot(kind="bar")

plt.title("Resume Distribution by Category")
plt.xlabel("Job Category")
plt.ylabel("Number of Resumes")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
missing_percentage = (
    jobs_df.isnull().sum() / len(jobs_df)
) * 100

print(missing_percentage.sort_values(ascending=False))
```

```python
print("Resume duplicates:", resume_df.duplicated().sum())

print("Job duplicates:", jobs_df.duplicated().sum())
```

## 3. Duplicate Record Investigation

Investigating repeated resume records to understand
their impact on the classification dataset.

```python
print("Total resume records:", len(resume_df))

print(
    "Exact duplicate rows:",
    resume_df.duplicated().sum()
)

print(
    "Unique complete records:",
    len(resume_df.drop_duplicates())
)
```
