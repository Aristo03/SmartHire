## 9. Skill Extraction

```python
from src.features.match_features import extract_skills, SKILL_ALIASES

print(len(SKILL_ALIASES), "skills in vocabulary")
print(extract_skills(
    "Experienced in Python, SQL, Power BI and machine learning. "
    "Built apps with ReactJS and Node.js. Knows C++, C# and JavaScript, not Java."
))
```

```python
jobs_clean["skills"] = jobs_clean["description"].apply(extract_skills)
resume_clean["skills"] = resume_clean["Resume"].apply(extract_skills)

print("Jobs with ≥1 skill:   ", round((jobs_clean["skills"].str.len() > 0).mean() * 100, 1), "%")
print("Resumes with ≥1 skill:", round((resume_clean["skills"].str.len() > 0).mean() * 100, 1), "%")
print("\nTop 20 skills in the job market:")
print(jobs_clean["skills"].explode().value_counts().head(20))
```

```python
jobs_clean.assign(skills=jobs_clean["skills"].str.join("|")).to_csv(config.JOBS_CLEAN, index=False)
resume_clean.assign(skills=resume_clean["skills"].str.join("|")).to_csv(config.RESUMES_CLEAN, index=False)
print("Saved ✅")
```
